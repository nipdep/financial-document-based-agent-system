## LM-Studio Python SDK User Manual
_Here we assume LMS Studio application running in background and as the user you have access to it's API_
you can test the API access by running 
```bash
curl https://127.0.0.1:1234/v1/models
```
in the terminal, then if you have access this command will return json object with all LLM models that has been downloaded to the server

````python
# init the library
import lmstudio as lms
````

````python
# see list of downloaded models : [resource : https://lmstudio.ai/docs/python/manage-models/list-downloaded ]
downloaded = lms.list_downloaded_models()
downloaded
````

````python
# see list of loaded models : [resource : https://lmstudio.ai/docs/python/manage-models/list-loaded ]
loaded = lms.list_loaded_models()
loaded
````

````python
# load model)
model = lms.llm("openai/gpt-oss-20b")
````

````python
result = model.respond("hi")
result
````

````python
# unload model
model.unload()
````

````python
# load embedding model
embed = lms.embedding_model("text-embedding-nomic-embed-text-v1.5")
````

````python
# generate embedding 
embedding = embed.embed("This is a sample text to be embedded.")
````

````python
embed.unload()
````

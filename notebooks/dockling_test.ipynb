```python
import requests
```

```python
# URL of your running Docling API
url = "http://localhost:8080/documents/convert"
params = {
    "extract_tables_as_images": "false",
    "image_resolution_scale": 4
}
```

```python
pdf_path = "../data/pog.pdf"


# Open the PDF file in binary mode
with open(pdf_path, "rb") as f:
    files = {"document": (pdf_path, f, "application/pdf")}
    
    # Make the POST request
    response = requests.post(url, params=params, files=files)
```

```python

# Check the status
if response.status_code == 200:
    data = response.json()  # Parsed JSON output
    print("Success! Parsed document:")
    print(data)
else:
    print(f"Request failed with status code {response.status_code}")
    print(response.text)
```

```python
if response.status_code == 200:
    data_text = response.json().get("markdown", "")
    output_md_path = "../data/pog_converted.md"
    with open(output_md_path, "w") as f:
        f.write(data_text)
```
